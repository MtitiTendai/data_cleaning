## Step 5 

```python
# Missing value(s) handling

for i in [' BMI ', 'Polio', 'Income composition of resources']:
    df[i] = df[i].fillna(df[i].median())
```

```python
df.isnull().sum()
```

```python
# Using KNNImputer

from sklearn.impute import KNNImputer

impute = KNNImputer() 
```

```python
for i in df.select_dtypes(include= 'number').columns:
    df[i]= impute.fit_transform(df[[i]])
```

```python
df.isnull().sum()
```

## Step 6

```python
# Outlier handling

def wisker(col):
    quartile1, quartile3 = np.percentile(col, [25, 75])
    inter_quartile_range = quartile3 - quartile1
    lower_wisker = quartile1 - 1.5 * inter_quartile_range
    upper_wisker = quartile3 + 1.5 * inter_quartile_range


    return lower_wisker, upper_wisker
```

```python
df.columns
```

```python
wisker(df['GDP'])
```

```python
for i in ['GDP','Total expenditure',' thinness  1-19 years', ' thinness 5-9 years']:
    lower_wisker, upper_wisker = wisker(df[i])
    df[i] = np.where(df[i] < lower_wisker, lower_wisker, df[i])
    df[i] = np.where(df[i] < upper_wisker, upper_wisker, df[i])
```

```python
for i in ['GDP','Total expenditure',' thinness  1-19 years', ' thinness 5-9 years']:
    sns.boxplot(df[i])
    plt.show()
```
